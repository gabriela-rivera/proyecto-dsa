## 2. Baseline (promedio móvil de 28 días)

## 3. Ridge: 6 valores de alpha

```python
import importlib
importlib.reload(mc)
```

```python
alphas = [0.01, 0.1, 1, 10, 100, 1000]
resultados = []
for a in alphas:
    r = mc.correr_modelo("ridge", f"ridge_alpha_{a}", Ridge(alpha=a),
                         {"alpha": a}, train, test, base)
    resultados.append(r)
```
