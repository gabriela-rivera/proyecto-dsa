```python
import sys, os
sys.path.append("../src")   # ajusta si tus notebooks usan otra ruta


import pandas as pd
from sklearn.ensemble import RandomForestRegressor
os.environ["MLFLOW_TRACKING_URI"] = "http://34.235.89.160:5000"
import modelos_comunes as mc

NOMBRE = "random_forest"
CONFIGS = [
    {"n_estimators": 200, "max_depth": 5,    "min_samples_leaf": 5},
    {"n_estimators": 200, "max_depth": 8,    "min_samples_leaf": 5},
    {"n_estimators": 300, "max_depth": 8,    "min_samples_leaf": 10},
    {"n_estimators": 300, "max_depth": None, "min_samples_leaf": 10},
    {"n_estimators": 500, "max_depth": 5,    "min_samples_leaf": 20},
    {"n_estimators": 500, "max_depth": 12,   "min_samples_leaf": 20},
]

mc.configurar_mlflow()
train, test = mc.cargar_datos()

# Métricas del baseline SIN crear otra run en MLflow
base = mc.calcular_metricas(test[mc.OBJETIVO].values, test["km_media28"].values)

resultados = []
for i, params in enumerate(CONFIGS, start=1):
    est = RandomForestRegressor(random_state=42, n_jobs=-1, **params)
    r = mc.correr_modelo(NOMBRE, f"rf_{i}", est, params, train, test, base)
    resultados.append(r)

df_res = pd.DataFrame(resultados).sort_values("cv_mae")
df_res[["run_name", "run_id", "n_estimators", "max_depth", "min_samples_leaf",
        "mae", "cv_mae", "mae_fecha_dias", "mejora_mae_pct"]]
```
