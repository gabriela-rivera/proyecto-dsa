**Objetivo:** Entrenar y evaluar modelos de regresión XGBoost para predecir `km_media_prox28`, utilizando las características y el pipeline de preprocesamiento definidos en el proyecto. Se experimentará con diferentes configuraciones de hiperparámetros, aplicando validación cruzada temporal y evaluando el desempeño mediante MAE, RMSE y MAE de fecha. Los experimentos se registrarán en MLflow.

```python
import sys, os
sys.path.append("../src")

import pandas as pd
from xgboost import XGBRegressor
import modelos_comunes as mc
```

```python
NOMBRE = "xgboost"

CONFIGS = [
    {"n_estimators": 200, "learning_rate": 0.03, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 400, "learning_rate": 0.03, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 400, "learning_rate": 0.05, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 200, "learning_rate": 0.03, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 400, "learning_rate": 0.03, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 400, "learning_rate": 0.05, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 1},

    {"n_estimators": 300, "learning_rate": 0.03, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 5},

    {"n_estimators": 300, "learning_rate": 0.05, "max_depth": 2,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 5},

    {"n_estimators": 300, "learning_rate": 0.03, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 5},

    {"n_estimators": 300, "learning_rate": 0.05, "max_depth": 3,
     "subsample": 0.8, "colsample_bytree": 0.8, "reg_lambda": 5},
]
```

```python
train, test = mc.cargar_datos()

base = mc.calcular_metricas(
    test[mc.OBJETIVO].values,
    test["km_media28"].values
)

print("Baseline:")
print(f"MAE: {base['mae']:.3f}")
print(f"RMSE: {base['rmse']:.3f}")
print(f"MAE fecha: {base['mae_fecha_dias']:.3f}")
```

```python
mc.AUTOR = "Gabriela Rivera"

mc.configurar_mlflow()
```

```python
resultados = []

for i, params in enumerate(CONFIGS, start=1):

    modelo = XGBRegressor(
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1,
        verbosity=0,
        **params
    )

    resultado = mc.correr_modelo(
        nombre=NOMBRE,
        run_name=f"xgb_{i}",
        estimador=modelo,
        params=params,
        train=train,
        test=test,
        base=base
    )

    resultados.append(resultado)

df_resultados = pd.DataFrame(resultados)
```
