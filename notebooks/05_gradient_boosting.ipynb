```python
import sys, os
sys.path.append("../src")

import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
import modelos_comunes as mc

NOMBRE = "gradient_boosting"
CONFIGS = [
    {"n_estimators": 100, "learning_rate": 0.05, "max_depth": 2, "subsample": 0.8},
    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 2, "subsample": 0.8},
    {"n_estimators": 200, "learning_rate": 0.05, "max_depth": 3, "subsample": 0.8},
    {"n_estimators": 300, "learning_rate": 0.03, "max_depth": 3, "subsample": 0.7},
    {"n_estimators": 300, "learning_rate": 0.10, "max_depth": 2, "subsample": 0.8},
    {"n_estimators": 500, "learning_rate": 0.02, "max_depth": 3, "subsample": 0.7},
]

mc.configurar_mlflow()
train, test = mc.cargar_datos()
base = mc.calcular_metricas(test[mc.OBJETIVO].values, test["km_media28"].values)

resultados = []
for i, params in enumerate(CONFIGS, start=1):
    est = GradientBoostingRegressor(random_state=42, **params)
    r = mc.correr_modelo(NOMBRE, f"gb_{i}", est, params, train, test, base)
    resultados.append(r)

df_res = pd.DataFrame(resultados).sort_values("cv_mae")
df_res[["run_name", "run_id", "n_estimators", "learning_rate", "max_depth",
        "mae", "cv_mae", "mae_fecha_dias", "mejora_mae_pct"]]
```
