# Notebook 2: Lasso
Objetivo: Entrenar y evaluar un modelo de regresión Lasso para predecir `km_media_prox28`, utilizando seis valores de regularización (`alpha`). Se calcularán las métricas del baseline para comparar el desempeño del modelo, sin registrar una nueva corrida para este.

```python
import sys
sys.path.append("../src")
import pandas as pd
from sklearn.linear_model import Lasso
import modelos_comunes as mc

mc.configurar_mlflow()
```

## 1. Datos

```python
train, test = mc.cargar_datos()
print("train:", train.shape, "| test:", test.shape)
print("version datos:", mc.version_datos())
```

## 2. Métricas del baseline (solo cálculo, sin registrar otra corrida)

```python
base = mc.calcular_metricas(test[mc.OBJETIVO].values, test["km_media28"].values)
print(f"Baseline -> MAE {base['mae']:.2f} km/dia | MAE fecha {base['mae_fecha_dias']:.2f} dias")
```

## 3. Lasso: 6 valores de alpha
En Ridge los mejores resultados estuvieron en alpha pequeño (0.01 a 0.1), así que la rejilla se concentra en valores bajos.

```python
alphas = [0.001, 0.01, 0.1, 0.5, 1, 5]
resultados = []
for a in alphas:
    r = mc.correr_modelo("lasso", f"lasso_alpha_{a}", Lasso(alpha=a, max_iter=20000),
                         {"alpha": a}, train, test, base)
    resultados.append(r)
```

```python
a = 5
r = mc.correr_modelo(
    "lasso",
    f"lasso_alpha_{a}",
    Lasso(alpha=a, max_iter=20000),
    {"alpha": a},
    train,
    test,
    base
)
resultados.append(r)
```

## 4. Resumen

```python
tabla = pd.DataFrame(resultados)[["run_name", "alpha", "cv_mae", "mae", "rmse",
                                  "mejora_mae_pct", "mae_fecha_dias", "mejora_fecha_pct"]]
tabla.sort_values("cv_mae").round(3)
```
