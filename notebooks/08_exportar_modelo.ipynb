# Para guardar el modelo en Joblib

```python
import sys
sys.path.append("../src")

import mlflow, joblib, sklearn
import modelos_comunes as mc

mlflow.set_tracking_uri("http://3.18.4.17:5000")
run_id = "0a718d0f35bf462eb5ba43de27190853"   # svr_8

train, test = mc.cargar_datos()

try:
    modelo = mlflow.sklearn.load_model(f"runs:/{run_id}/model")
    print("Cargado desde MLflow")
except Exception as e:
    print("No hay modelo guardado:", e)
    print(mlflow.get_run(run_id).data.params)
```

```python
import numpy as np

# 1. Debe predecir con las columnas crudas (vehiculo y contrato como texto)
X = test[mc.FEATURES]
pred = modelo.predict(X)
print("predicciones:", pred[:5])

# 2. Comprueba que es el mismo modelo: el MAE debe dar ~15.38
mae = np.mean(np.abs(test[mc.OBJETIVO] - pred))
print("MAE test:", round(mae, 3), "(esperado: 15.383)")

# 3. Mira qué hay dentro
print(type(modelo))
print(modelo)

# 4. Guarda
joblib.dump(modelo, "../modelo_final.joblib")
print("Guardado. sklearn:", sklearn.__version__)
```
