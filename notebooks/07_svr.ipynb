# Notebook 7: SVR
**Objetivo:** Entrenar y evaluar modelos de regresión SVR (Support Vector Regression) con kernel RBF para predecir `km_media_prox28`, utilizando las características y el pipeline de preprocesamiento definidos en el proyecto. Se experimentará con diferentes configuraciones de hiperparámetros, aplicando validación cruzada temporal y evaluando el desempeño mediante MAE, RMSE y MAE de fecha. Los experimentos se registrarán en MLflow.

```python
import sys, os
sys.path.append("../src")

import pandas as pd
from sklearn.svm import SVR
import modelos_comunes as mc
```

```python
NOMBRE = "svr"

CONFIGS = [
    {"C": 0.1, "epsilon": 0.1, "gamma": "scale"},
    {"C": 1, "epsilon": 0.1, "gamma": "scale"},
    {"C": 10, "epsilon": 0.1, "gamma": "scale"},
    {"C": 100, "epsilon": 0.1, "gamma": "scale"},
    {"C": 1, "epsilon": 0.5, "gamma": "scale"},
    {"C": 10, "epsilon": 0.5, "gamma": "scale"},
    {"C": 1, "epsilon": 0.1, "gamma": 0.01},
    {"C": 10, "epsilon": 0.1, "gamma": 0.01},
]
```

```python
mc.TRACKING_URI = "http://3.88.204.210:5000"
mc.AUTOR = "Gabriela Rivera"

mc.configurar_mlflow()
```

```python
train, test = mc.cargar_datos()

base = mc.calcular_metricas(
    test[mc.OBJETIVO].values,
    test["km_media28"].values
)

print("Baseline:")
print(f"MAE: {base['mae']:.3f}")
print(f"RMSE: {base['rmse']:.3f}")
print(f"MAE fecha: {base['mae_fecha_dias']:.3f}")
```

```python
resultados = []

for i, params in enumerate(CONFIGS, start=1):

    modelo = SVR(
        kernel="rbf",
        **params
    )

    resultado = mc.correr_modelo(
        nombre=NOMBRE,
        run_name=f"svr_{i}",
        estimador=modelo,
        params=params,
        train=train,
        test=test,
        base=base
    )

    resultados.append(resultado)

df_resultados = pd.DataFrame(resultados)
```

```python
df_resultados = df_resultados.sort_values("cv_mae").reset_index(drop=True)

df_resultados[
    [
        "run_name",
        "C",
        "epsilon",
        "gamma",
        "mae",
        "rmse",
        "cv_mae",
        "mae_fecha_dias",
        "mejora_mae_pct"
    ]
]
```
